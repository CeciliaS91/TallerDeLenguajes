```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os 
import kagglehub

# Download latest version
path = kagglehub.dataset_download("crystalbaby/global-programming-languages-intelligence-dataset")

print("Path to dataset files:", path)

archivo_csv = os.path.join(path, "global_programming_languages_dataset_2020_2026.csv ")
```

```python
df = pd.read_csv(archivo_csv)
print(df.head()) 
```

```python
#Top ten rating según web development score 
plt.figure(figsize=(12,6))
sns.barplot(
    data=df.head(10),
    x="language",
    y="web_development_score",
    hue="language",       # asignamos hue
    palette="viridis",
    legend=False          # evita mostrar leyenda duplicada
)
plt.title("Top 10 Lenguajes por Rating")
plt.xticks(rotation=45)
plt.show()
```

```python
# Heatmap de correlaciones entre años desde 2020 a 2026, el puntaje de performance y el puntaje de demanda 
plt.figure(figsize=(8,6))
sns.heatmap(df[['performance_score','demand_score','year']].corr(), annot=True, cmap="coolwarm")
plt.title("Correlaciones entre variables")
plt.show()
```

```python
# Insights (conocimientos que aportan valor)
top_growth = df.sort_values("demand_score", ascending=False).head(5)
top_decline = df.sort_values("demand_score").head(5)

print("\nLenguajes con mayor crecimiento:\n", top_growth[['language','demand_score']])
print("\nLenguajes con mayor caída:\n", top_decline[['language','demand_score']])
```
